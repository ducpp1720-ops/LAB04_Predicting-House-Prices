# Lab 04 — California Housing EDA + PyTorch MLP Regression

EDA required by the lab: shape, info, missing values, duplicates, descriptive statistics, distributions, correlation and outlier inspection.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
train_df = pd.read_csv(ROOT / 'data/raw/california_housing_train.csv')
test_df = pd.read_csv(ROOT / 'data/raw/california_housing_test.csv')
display(train_df.head())
```

## 1. EDA

```python
print('Train shape:', train_df.shape)
print('Test shape:', test_df.shape)
train_df.info()
print('\nMissing values:')
print(train_df.isna().sum())
print('\nDuplicate rows:', train_df.duplicated().sum())
display(train_df.describe().T)
```

## 2. Distributions and outliers

```python
train_df.hist(figsize=(14,10), bins=30)
plt.tight_layout()
plt.show()
plt.figure(figsize=(10,6))
sns.boxplot(data=train_df[['total_rooms','total_bedrooms','population','households','median_income']])
plt.xticks(rotation=20)
plt.title('Boxplots for Outlier Inspection')
plt.tight_layout()
plt.show()
```

## 3. Correlation

```python
plt.figure(figsize=(10,8))
sns.heatmap(train_df.corr(numeric_only=True), cmap='coolwarm', center=0)
plt.title('Feature Correlation Matrix')
plt.tight_layout()
plt.show()
```

## 4. Preprocessing

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
features = ['longitude','latitude','housing_median_age','total_rooms','total_bedrooms','population','households','median_income']
target = 'median_house_value'
X_train, X_val, y_train, y_val = train_test_split(train_df[features], train_df[target], test_size=0.2, random_state=42)
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_val_scaled = scaler.transform(X_val)
X_test_scaled = scaler.transform(test_df[features])
print(X_train_scaled.shape, X_val_scaled.shape, X_test_scaled.shape)
```
